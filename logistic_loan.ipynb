```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("loan_approval_data.csv")
df_copy=df.copy()

# df.describe()
```

```python
### handling missing values
df_copy["Applicant_Income"]=df_copy["Applicant_Income"].fillna(df_copy["Applicant_Income"].mean()).copy()
df_copy["Coapplicant_Income"]=df_copy["Coapplicant_Income"].fillna(df_copy["Coapplicant_Income"].mean()).copy()
df_copy["Age"]=df_copy["Age"].fillna(df_copy["Age"].mean()).copy()
df_copy["Dependents"]=df_copy["Dependents"].fillna(df_copy["Dependents"].mean()).copy()
df_copy["Existing_Loans"]=df_copy["Existing_Loans"].fillna(df_copy["Existing_Loans"].mean()).copy()
df_copy["DTI_Ratio"]=df_copy["DTI_Ratio"].fillna(df_copy["DTI_Ratio"].mean()).copy()
df_copy["Savings"]=df_copy["Savings"].fillna(df_copy["Savings"].mean()).copy()
df_copy["Collateral_Value"]=df_copy["Collateral_Value"].fillna(df_copy["Collateral_Value"].mean()).copy()
df_copy["Loan_Amount"]=df_copy["Loan_Amount"].fillna(df_copy["Loan_Amount"].mean()).copy()
df_copy["Loan_Term"] = df["Loan_Term"].fillna(df["Loan_Term"].mean())
df_copy["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].mean())
```

```python
df_copy["Employment_Status"]=df_copy["Employment_Status"].fillna(df_copy["Employment_Status"].mode()[0]).copy()
df_copy["Marital_Status"]=df_copy["Marital_Status"].fillna(df_copy["Marital_Status"].mode()[0]).copy()
df_copy["Loan_Purpose"]=df_copy["Loan_Purpose"].fillna(df_copy["Loan_Purpose"].mode()[0]).copy()
df_copy["Property_Area"]=df_copy["Property_Area"].fillna(df_copy["Property_Area"].mode()[0]).copy()
df_copy["Education_Level"]=df_copy["Education_Level"].fillna(df_copy["Education_Level"].mode()[0]).copy()
df_copy["Gender"]=df_copy["Gender"].fillna(df_copy["Gender"].mode()[0]).copy()
df_copy["Employer_Category"]=df_copy["Employer_Category"].fillna(df_copy["Employer_Category"].mode()[0]).copy()
df_copy["Loan_Approved"]=df_copy["Loan_Approved"].fillna(df_copy["Loan_Approved"].mode()[0]).copy()
```

```python
df_copy=pd.get_dummies(df_copy,columns=["Employment_Status","Marital_Status","Loan_Purpose","Property_Area","Education_Level","Gender","Employer_Category"],drop_first=True)
```

```python
X=df_copy.drop(["Applicant_ID","Loan_Approved"],axis=1).copy()
Y=df_copy["Loan_Approved"].copy()
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print(x_train.columns.tolist())
```

```python
from sklearn.preprocessing import StandardScaler
Scaler=StandardScaler()
x_trains=Scaler.fit_transform(x_train)
x_tests=Scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,f1_score,recall_score
from sklearn.model_selection import GridSearchCV
model=LogisticRegression()
param_grid={"max_iter":[1000]}
for c in [3,5,8, 10, 100]:
   m=GridSearchCV(
      model,
      param_grid,
      cv=c
   )
   m.fit(x_trains,y_train) 
   y_pred=m.predict(x_tests)
   print(accuracy_score(y_test,y_pred)*100)
   print("Precision:", precision_score(y_test, y_pred, pos_label="Yes") * 100)
   print("Recall   :", recall_score(y_test, y_pred, pos_label="Yes") * 100)
   print("F1 Score :", f1_score(y_test, y_pred, pos_label="Yes") * 100)
```

```python
from sklearn.pipeline import Pipeline
pipe=Pipeline([('scelar',StandardScaler()),("model",LogisticRegression())])
param_grid={"model__max_iter":[1000]}
model=GridSearchCV(
      pipe,
      param_grid,
      cv=5
   )
model.fit(x_trains,y_train)
y_pred=model.predict(x_tests)
print("Accuracy:",accuracy_score(y_test,y_pred)*100)
print("Precision:", precision_score(y_test, y_pred, pos_label="Yes") * 100)
print("Recall   :", recall_score(y_test, y_pred, pos_label="Yes") * 100)
print("F1 Score :", f1_score(y_test, y_pred, pos_label="Yes") * 100)
```
